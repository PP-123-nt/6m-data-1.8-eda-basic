### Challenge 2: Clean the Data

**Tasks:**
1. Convert `revenue_raw` into a numeric column called `revenue_sgd`.
2. Replace the fake blanks in `notes` (`""`, `n/a`, `N.A.`, `-`) with `NaN`.
3. Mask the impossible values: negative takings, takings above 20,000, and ticket counts of zero or
   less.
4. Fill the resulting holes in `revenue_sgd` and `tickets` with each column's median — **in the right
   order**.
5. Standardise `outlet` to the four real café names, and `daypart` to Morning / Midday / Evening.
6. Verify: no impossible values left, and `outlet` has exactly four values.

```python
clean_may = may.copy()  # copy another version clean_may

# 1. converrt revenue into numeric

clean_may["revenue_sgd"] = pd.to_numeric(
    clean_may["revenue_raw"].str.replace(r"[^0-9.\-]", "", regex=True), errors="coerce"
)

display(may.head())  # may record remain unchanged

display(clean_may)   # cleaning done in clean_may version
```

```python
# 2. replae notes fake blanks -> real blanks (NaN), so .isna() tells the truth from here on.

clean_may["notes"] = clean_may["notes"].replace(["", "n/a", "N.A.", "-"], np.nan)

clean_may
```

```python
# 3. mask the invalid values (find -> decide -> apply), revenue > 20,000 & ticket <= 0

clean_may["revenue_sgd"] = clean_may["revenue_sgd"].mask(
    (clean_may["revenue_sgd"] < 0) | (clean_may["revenue_sgd"] > 20000)
)
clean_may["tickets"] = clean_may["tickets"].mask(clean_may["tickets"] <= 0)

clean_may

# alternate method using where
# clean_may["revenue_sgd"] = clean_may["revenue_sgd"].where(clean_may["revenue_sgd"].between(0, 20000))
# clean_may["tickets"] = clean_may["tickets"].where(clean_may["tickets"] > 0)
```

```python
# 4. compute the medians and update NaN with median 
# replace invalid value to NaN before median computation to ensure accuracy

clean_may = clean_may.fillna({
    "revenue_sgd": clean_may["revenue_sgd"].median(),
    "tickets": clean_may["tickets"].median(),
    "staff_on_shift": clean_may["staff_on_shift"].median(),
})

clean_may
```

```python
# 5. standardize outlet and daypart naming

outlet_map = {
    "Raffles Place": "Raffles Place", "raffles place": "Raffles Place",
    "RAFFLES PLACE": "Raffles Place", "Raffles Pl.": "Raffles Place",
    "Marina Bay": "Marina Bay", "marina bay": "Marina Bay", "Marina Bay ": "Marina Bay",
    "Tampines Mall": "Tampines Mall", "tampines mall": "Tampines Mall",
    "Holland Village": "Holland Village", "Holland V": "Holland Village",
}
daypart_map = {
    "Morning": "Morning", "morning": "Morning", "AM": "Morning",
    "Midday": "Midday", "midday": "Midday", "Lunch": "Midday",
    "Evening": "Evening", "evening": "Evening", "PM": "Evening",
}
clean_may["outlet_name"] = clean_may["outlet"].map(outlet_map)
clean_may["daypart"] = clean_may["daypart"].map(daypart_map)

clean_may
```

```python
# 6. verify cleaning

print("unmapped outlets:", clean_may["outlet_name"].isna().sum())
print("unmapped dayparts:", clean_may["daypart"].isna().sum())
print("cafés:", clean_may["outlet_name"].nunique())
print("min revenue:", clean_may["revenue_sgd"].min(), "| max:", clean_may["revenue_sgd"].max())
print("min tickets:", clean_may["tickets"].min())
```

### Challenge 3: Transform & Summarise

**Tasks:**
1. Convert `date_text` to a real datetime column called `date`. Careful: `01/05/2025` is 1 May here.
2. Clean the `manager` column so each manager's name appears exactly once, in Title Case.
3. Add a `shift_size` column with `pd.cut`: 0–350 `"Quiet"`, 350–700 `"Normal"`, 700–1500 `"Busy"`.
4. Build a summary table: one row per café, with total revenue, number of shifts and average revenue
   per shift, sorted by total revenue.

```python
# 1. dayfirst=True matters: without it, 01/05/2025 could be read as 5 January.

clean_may["date"] = pd.to_datetime(clean_may["date_text"], dayfirst=True)  # dayfirst=True -> 1st number is day

clean_may[['date_text', 'date']].head()
```

```python
# 2. strip the spaces off the ENDS, squeeze the doubled spaces in the MIDDLE, then Title Case.
#    `\s+` means "one or more whitespace characters"; replacing them with a single space turns
#    "Priya  Nair" into "Priya Nair". `.str.strip()` alone would not have caught that one.

clean_may["manager"] = (
    clean_may["manager"].str.strip().str.replace(r"\s+", " ", regex=True).str.title()
)
print(clean_may["manager"].unique())
```

```python
# 3. group into 3 bin size

clean_may["shift_size"] = pd.cut(
    clean_may["revenue_sgd"], bins=[0, 350, 700, 1500], labels=["Quiet", "Normal", "Busy"]
)
print(clean_may["shift_size"].value_counts())
```

```python
# 4. summary table by café, with total revenue, number of shifts and average revenue per shift, sorted by total revenue.

may_summary = clean_may.groupby("outlet_name", observed=True).agg(
    revenue=("revenue_sgd", "sum"),
    shifts=("revenue_sgd", "size"),
    avg_shift=("revenue_sgd", "mean"),
).round(2).sort_values("revenue", ascending=False)

may_summary
```

Do the same audit on the **real** file, `data/cafe_june_raw.csv`, but for one outlet only:

```python
import pandas as pd
import numpy as np

# assignment-working location is same level as data folder, no need use "../data/cafe_june_raw.csv" to go 1 level up
june_raw = pd.read_csv("data/cafe_june_raw.csv") 

june_raw
```

```python
# 1. filter marina bay outlets records and copy into a new file

display(june_raw["outlet"].value_counts())

# marina_outlet is a new variable created w/o adding new column to june_raw file
# .str.strip() ignore hidden space in marina bay naming
# .str.lower() ignore upper & lower cap variants (MARINA Bay, marina Bay etc)
marina_outlet = june_raw["outlet"].str.strip().str.lower() == "marina bay" 

# june_raw[marina_outlet] is boolean indexing tells pandas to keep the rows in june_raw where outlet = marina bay
marina = june_raw[marina_outlet].copy()

print("rows:", len(marina), "| spellings matched:", marina["outlet"].nunique())

#print("rows:", len(marina), "| Standardized spellings:", marina["outlet"].str.strip().str.lower().unique().tolist())

marina
```

```python
marina["revenue_sgd"] = pd.to_numeric(
    marina["revenue_raw"].str.replace(r"[^0-9.\-]", "", regex=True), errors="coerce"
)

bad_revenue_condition = (marina["revenue_sgd"] < 0) | (marina["revenue_sgd"] > 20000)

# display the rows that meet the condition for verification
display(marina[bad_revenue_condition])

# display duplicates
all_duplicates = marina[marina.duplicated(keep=False)] # show all duplicate rows
display(all_duplicates)

# marina[marina.duplicated()] -> only show duplicate once
```

```python
# 2. cleaning and compute jun revenue
# remove 3 duplicate records, replace invalid revenue values to median, standardize date format

marina["revenue_sgd"] = pd.to_numeric(
    marina["revenue_raw"].str.replace(r"[^0-9.\-]", "", regex=True), errors="coerce"
)
marina = marina.drop_duplicates()
marina["revenue_sgd"] = marina["revenue_sgd"].mask(
    (marina["revenue_sgd"] < 0) | (marina["revenue_sgd"] > 20000)
)
marina["revenue_sgd"] = marina["revenue_sgd"].fillna(marina["revenue_sgd"].median())
marina["date"] = pd.to_datetime(marina["date_text"], dayfirst=True)

print(f"Marina Bay, June 2025: ${marina['revenue_sgd'].sum():,.2f} across {len(marina)} shifts")

marina
```

```python
# 3. export result to csv

june_mb = marina[[
    "date", "daypart", "tickets", "revenue_sgd",
]].sort_values(["date", "daypart"])

june_mb.to_csv("data/june_mb.csv", index=False)

june_mb.head()
```
