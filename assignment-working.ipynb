### Challenge 2: Clean the Data

**Tasks:**
1. Convert `revenue_raw` into a numeric column called `revenue_sgd`.
2. Replace the fake blanks in `notes` (`""`, `n/a`, `N.A.`, `-`) with `NaN`.
3. Mask the impossible values: negative takings, takings above 20,000, and ticket counts of zero or
   less.
4. Fill the resulting holes in `revenue_sgd` and `tickets` with each column's median — **in the right
   order**.
5. Standardise `outlet` to the four real café names, and `daypart` to Morning / Midday / Evening.
6. Verify: no impossible values left, and `outlet` has exactly four values.

```python
clean_may = may.copy()  # copy another version clean_may

# 1. converrt revenue into numeric

clean_may["revenue_sgd"] = pd.to_numeric(
    clean_may["revenue_raw"].str.replace(r"[^0-9.\-]", "", regex=True), errors="coerce"
)

display(may.head())  # may record remain unchanged

display(clean_may)   # cleaning done in clean_may version
```

```python
# 2. replae notes fake blanks -> real blanks (NaN), so .isna() tells the truth from here on.

clean_may["notes"] = clean_may["notes"].replace(["", "n/a", "N.A.", "-"], np.nan)

clean_may
```

```python
# 3. mask the invalid values (find -> decide -> apply), revenue > 20,000 & ticket <= 0

clean_may["revenue_sgd"] = clean_may["revenue_sgd"].mask(
    (clean_may["revenue_sgd"] < 0) | (clean_may["revenue_sgd"] > 20000)
)
clean_may["tickets"] = clean_may["tickets"].mask(clean_may["tickets"] <= 0)

clean_may

# alternate method using where
# clean_may["revenue_sgd"] = clean_may["revenue_sgd"].where(clean_may["revenue_sgd"].between(0, 20000))
# clean_may["tickets"] = clean_may["tickets"].where(clean_may["tickets"] > 0)
```

```python
# 4. compute the medians and update NaN with median 
# replace invalid value to NaN before median computation to ensure accuracy

clean_may = clean_may.fillna({
    "revenue_sgd": clean_may["revenue_sgd"].median(),
    "tickets": clean_may["tickets"].median(),
    "staff_on_shift": clean_may["staff_on_shift"].median(),
})

clean_may
```

```python
# 5. standardize outlet and daypart naming

outlet_map = {
    "Raffles Place": "Raffles Place", "raffles place": "Raffles Place",
    "RAFFLES PLACE": "Raffles Place", "Raffles Pl.": "Raffles Place",
    "Marina Bay": "Marina Bay", "marina bay": "Marina Bay", "Marina Bay ": "Marina Bay",
    "Tampines Mall": "Tampines Mall", "tampines mall": "Tampines Mall",
    "Holland Village": "Holland Village", "Holland V": "Holland Village",
}
daypart_map = {
    "Morning": "Morning", "morning": "Morning", "AM": "Morning",
    "Midday": "Midday", "midday": "Midday", "Lunch": "Midday",
    "Evening": "Evening", "evening": "Evening", "PM": "Evening",
}
clean_may["outlet_name"] = clean_may["outlet"].map(outlet_map)
clean_may["daypart"] = clean_may["daypart"].map(daypart_map)

clean_may
```

```python
# 6. verify cleaning

print("unmapped outlets:", clean_may["outlet_name"].isna().sum())
print("unmapped dayparts:", clean_may["daypart"].isna().sum())
print("cafés:", clean_may["outlet_name"].nunique())
print("min revenue:", clean_may["revenue_sgd"].min(), "| max:", clean_may["revenue_sgd"].max())
print("min tickets:", clean_may["tickets"].min())
```
